```python
import numpy as np
import pandas as pd

df = pd.DataFrame({"name": ["tom", "nick", "raj", "harsh"],
                   "age": [23, 34, 54, 36],
                   "salary": [10000, 20000, 50000, 45000],
                   "city ": ["delhi", "mumbai", "pune", "banglore"],
                   "department": ["IT", "HR","Finane","sales"]})
print(df)
print(df.isna().sum())

df =df. dropna(axis = 0)
df.dropna(inplace = True)

df.fillna(0, inplace = True)
df ["salary"]=df["salary"].fillna(0)

print(df)
df.info()
print(df.duplicated())
df.drop_duplicates(inplace = True)

print(df)
df.info()

print(df.duplicated())

df.drop_duplicates(inplace = True)

print(df)
df_group = df.groupby("department").agg(
    {"salary": ["mean", "sum", "max", "min"]
     }
)

print(df_group)
df.to_csv("demo.csv",index= False)

df.to_excel("demo.xlsx",index= False)

```
