```python

import numpy as np
import matplotlib.pyplot as plt

# Create figure
fig = plt.figure()



ax1 = fig.add_subplot(121, projection='3d')

t = np.linspace(0, 2 * np.pi, 100)

x = np.cos(t)
y = np.sin(t)
z = t

ax1.plot(x, y, z)

ax2 = fig.add_subplot(122, projection='3d')

x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)

X, Y = np.meshgrid(x, y)

Z = np.sin(X) * np.cos(Y)

ax2.plot_surface(X, Y, Z, cmap='jet')


plt.tight_layout()
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

# Step 1: Create figure
fig = plt.figure(figsize=(10, 7))

# Step 2: Create 3D axes
ax = fig.add_subplot(111, projection='3d')

# Step 3: Generate X and Y values
x = np.linspace(-5, 5, 50)
y = np.linspace(-5, 5, 50)

# Step 4: Create coordinate grid
x, y = np.meshgrid(x, y)

# Step 5: Calculate Z values
z = x**2 + y**2

# Step 6: Create 3D wireframe
ax.plot_wireframe(
    x, y, z,
    cmap='jet',
    linewidth=0.7
)

# Step 7: Display graph
plt.show()
```
