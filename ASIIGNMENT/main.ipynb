```python

import numpy as np
import pandas as pd

df = pd.read_csv("sales_data.csv")
print(df)

df_category = df.groupby("Category")["TotalSales"].sum().reset_index()
print(df_category)
df_category.to_excel("category_sales.xlsx", index=False)

df_region = df.groupby("Region")["TotalSales"].sum().reset_index()
print(df_region)
df_region.to_excel("region_sales.xlsx", index=False)

df_salesperson = df.groupby("SalesPerson")["TotalSales"].sum().reset_index()
print(df_salesperson)
df_salesperson.to_excel("salesperson_sales.xlsx", index=False)


df_product = df.groupby("ProductName")["TotalSales"].sum().reset_index()
print(df_product)
df_product.to_excel("product_sales.xlsx", index=False)


df_quantity = df.groupby("ProductName")["Quantity"].sum().reset_index()
print(df_quantity)
df_quantity.to_excel("product_quantity.xlsx", index=False)


```
