```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os

df = pd.read_csv(os.path.join(path, "mflix.comments.csv"))

# 1. Check missing values
print(df.isna())
print(df.isna().sum())

# 2. Check DataFrame information
df.info()

# 3. Remove duplicate rows
df = df.drop_duplicates()

# 4. View data
print(df.head())
print(df.tail())

# 5. Find top 10 commenters by number of comments
top_actors = (
    df.groupby(["name", "email", "_id","movie_id"])
      .size()
      .nlargest(10)
      .reset_index(name="count")
)


plt.bar(top_actors["name"], top_actors["count"])
plt.plot(top_actors["name"],top_actors["movie_id"])

plt.show()


```
