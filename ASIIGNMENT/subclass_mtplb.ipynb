```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("../datasets/amazon.csv")

print(df.head(100))
print(df.isna().sum())
df.info()

df = df.drop_duplicates()

print(df.describe())

# Count products by category
top_categories = df["category"].value_counts().head(10)

# Bar chart
plt.bar(top_categories.index, top_categories.values)

plt.xlabel("Category")
plt.ylabel("Number of Products")
plt.title("Top 10 Product Categories")

plt.xticks(rotation=90, ha="right")
plt.tight_layout()
plt.show()

top_categories = df["category"].value_counts().tail(10)
plt.barh(top_categories.index, top_categories.values)
plt.xlabel("Number of Products")
plt.ylabel("Category")
plt.title("Bottom 10 Product Categories")
plt.tight_layout()
plt.show()





```
